# tkh-project-3-unsupervised_learning - Transform notebook

```python
import pandas as pd
```

## Load dataset

Here we see that we 24 columns in dataset and no missing values.

```python
df_orig = pd.read_csv('../data/train.csv')
df_orig.info()
```

## Rename columns, drop index column

```python
columns_orig = list(df_orig.columns)
columns_orig
```

```python
df = df_orig.copy()

# Drop the index column
df.drop('Unnamed: 0', axis=1, inplace=True)

# Rename columns
columns_new = [
    'artist-name',
    'track-name',
    'release-date',
    'genre',
    'lyrics',
    'len',
    'dating',
    'violence',
    'world-life',
    'night-time',
    'shake_audience',
    'family-gospel',
    'romantic',
    'communication',
    'obscene',
    'music',
    'movement-places',
    'light-visual',
    'family-spiritual',
    'sadness',
    'feelings',
    'topic',
    'age',
]
df.columns = columns_new

# Sort columns
df = df.reindex(sorted(columns_new), axis=1)
```

## Dummy encode the desireable categorical columns

```python
df_dummies = pd.get_dummies(df[['genre', 'topic']], drop_first=True)
df_dummies
```

## Drop unwanted columns + combine with the dummy encoded ones

```python
df_final = df.copy()

# Drop the unwanted columns and the original, non-encoded columns
df_final.drop(['genre', 'topic', 'artist-name', 'track-name', 'lyrics'], axis=1, inplace=True)

# Combine encoded columns with remaining columns
df_final = pd.concat([df_final, df_dummies], axis=1)

# Sort columns alphabetically
df_final = df_final.reindex(sorted(df_final.columns), axis=1)
df_final
```

```python
df_final.info()
```

## Save dataframe to disk

```python
df_final.to_csv('../data/formatted.csv')
```
