```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

## Load dataset

Here we see that we 24 columns in dataset and no missing values.

```python
df_orig = pd.read_csv('../data/train.csv')
df_orig.info()
```

## Rename columns, drop index column

```python
columns_orig = list(df_orig.columns)
columns_orig
```

```python
df = df_orig.copy()

# Drop the index column
df.drop('Unnamed: 0', axis=1, inplace=True)

# Rename columns
columns_new = [
    'artist-name',
    'track-name',
    'release-date',
    'genre',
    'lyrics',
    'len',
    'dating',
    'violence',
    'world-life',
    'night-time',
    'shake_audience',
    'family-gospel',
    'romantic',
    'communication',
    'obscene',
    'music',
    'movement-places',
    'light-visual',
    'family-spiritual',
    'sadness',
    'feelings',
    'topic',
    'age',
]
df.columns = columns_new

# Sort columns
df = df.reindex(sorted(columns_new), axis=1)
```

## Numeric columns

Let's examine our numeric columns.

```python
df_num = df.select_dtypes(include=['number'])

numeric_columns = list(df_num.columns)
display(numeric_columns)

print('Number of num columns:', len(numeric_columns))
```

As we can see from stats in the below table, only two of our features fit into the range (0, 1), and they are `len` and `release date`. `len` is the number of words in a song's lyrics and `release_date` is a year. `len` ranges from 1 to 199 words, and  `release_date` ranges from 1950 to 2019. Of the other features, `age` has highest standard deviation, 0.264, and the `shake_audience` has the lowest standard deviation, 0.041.

```python
df_num.describe()
```

Below, we plot histograms of the 18 numeric features. All but three of the features (`age`, `len`, and `release-date`) are heavily right skewed with a steep dropoff. `age` and `release-date` are somewhat uniformly distributed, with a drop off for newer values of `age` and older values of `release-date`. That inverse relationship is a little counter-intuitive, but the uniform tendency is desireable because we want a balanced variety of music periods. `len` mostly follows a normal distribution. That distribution makes sense because extremely terse and extremely length lyrics are less common, assuming were sampling non-instrumental music. Given that most modern music tracks are between 3 - 4.5 minute in duration, it seems plausable to have a frequently occurring mean length of song lyrics.

```python
# Plot histograms
fig, axes = plt.subplots(5, 4, figsize=(20, 20))
for col, ax in zip(numeric_columns, axes.ravel()):
    _ = df_num[col].plot.hist(ax=ax)
    ax.set_title(col)
```

In the correlation plot below, we see two perfectly and negatively correlated features--`age` and `release-date`. That makes sense because the two features mean the same thing, just in inverse directions: higher `age` value means older music and higher `release-date` values means newer music. The only two other correlations that stand out are a medium correlation between `obscene` and `len` and a small correlation between `release-date` and `len`. Those two correlations aren't so easily deciphered. The rest of the features have quite low correlations.

```python
# Plot a correlation heat map
corr = df_num.corr()
plt.figure(figsize=(12, 10))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0, square=True)
plt.title('Correlation Heatmap')
```
